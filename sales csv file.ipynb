```python
import pandas as jl
import matplotlib.pyplot as plt

df = jl.read_csv("C:/Users/GCS-3.3/Desktop/sales.csv.html")

plt.plot(df["Month"], df["Sales"], marker="o")

plt.title("Sales Over Months")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("C:/Users/GCS-3.3/Desktop/sales.csv.html")

plt.bar(df["Month"], df["Profit"])

plt.title("Profit by Month")
plt.xlabel("Month")
plt.ylabel("Profit")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("C:/Users/GCS-3.3/Desktop/sales.csv.html")

plt.pie(df["Sales"], labels=df["Month"], autopct="%1.1f%%")

plt.title("Monthly Sales Share")

plt.show()
```
